### Weighted mean

```python
temp_df = pd.DataFrame({'col1':[1,2,3],'col2':[10,20,30]})
```

```python
temp_df
```

```python
temp_df['col2'].mean()
```

### How to use numpy to do the weighted mean

```python
np.average(temp_df['col2'])
```

### Or Weighted mean if you want

```python
np.average(temp_df['col2'], weights= temp_df['col1'])
```

### Skewness
In cases where you can't tell visually what is going on. You can calculate skewness 

```python
df['age'].plot.hist()
```

```python
df['age'].skew()
```

```python
df['hours-per-week'].plot.hist()
```

```python
df['hours-per-week'].skew()
```

```python
df.columns
```

If `df['column'].skew()` gives you a positive number it is rightly or positively skewed. Else negative/left skewed 

### Kurotsis 

```python
df['age'].kurt()
```

```python
df['capital-loss'].kurt()
```

```python
df['age'].plot.hist(bins=10)
```

```python
df['capital-loss'].plot.hist(bins=10)
```

### Tips to remember 

Skewness tells you whether the data is on the right/left or center and by how many degrees.

Kurtosis tells you whether the data is concentrated or spread out.

```python
import time
from IPython.display import clear_output

def timer(seconds):
    for i in range(seconds, 0, -1):
        clear_output(wait=True)
        print(f"Time left: {i} seconds")
        time.sleep(1) 
    print("Time's up!") 
```
