##### Convert date coloumn to proper format

```python
data['Order Date'] = pd.to_datetime(data['Order Date'])
```

```python
data.info()
```

```python
data['Ship Date'] = pd.to_datetime(data['Ship Date'])
```

```python
data.info()
```

##### Add 3 new columns 1 for year, 2 for month, 3 for week

```python
data['Order Day of Week'] = data['Order Date'].dt.dayofweek
data['Order Month'] = data['Order Date'].dt.month
data['Order Year'] = data['Order Date'].dt.year
```

```python
data.head()
```

##### Monthly sales analysis

```python
sales_by_month = data.groupby('Order Month')['Sales'].sum().reset_index()
                    #did group by so we can find monthly sales
                    #for monthly sales we need order month and the sales
```

```python
sales_by_month
#it shows the sales by month
```

```python
fig = px.line(sales_by_month,
              x = 'Order Month',
              y = 'Sales',
              title = 'Monthly Sales Analysis')

fig.show()
```

##### Sales by category

```python
sales_by_category = data.groupby('Category')['Sales'].sum().reset_index()
```

```python
sales_by_category
```

```python
fig = px.pie(sales_by_category,
             values = 'Sales',
             names = 'Category',
             hole = 0.4,
             color_discrete_sequence = px.colors.qualitative.Pastel)

fig.update_traces(textposition = 'inside',
                  textinfo = 'label + percent')

fig.update_layout(title_text = 'Sales Analysis by Category',
                  title_font = dict(size = 20))

fig.show()
```
