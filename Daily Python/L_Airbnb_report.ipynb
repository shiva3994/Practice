##### Handling missing values

```python
df['last review'] = pd.to_datetime(df['last review'],errors='coerce')
#Select columns for change
#Tell panda we wish to change it 'last review' to date time instead of str
#Errors = 'coerce' tells that if there is missing value it should is be considered as NAT format (Not a thing)
```

```python
df.head()
```

```python
df.info()
#To check the data type of 'last review'
```

```python
#Fill 'reviews per month' with 0 and replace 'last review' with minimum value of last review
df.fillna({'reviews per month' : 0, 'last review' : df['last review'].min()}, inplace=True)
```

```python
#Drop na from selected columns
df.dropna(subset = ['NAME','host name'], inplace = True)
```

```python
print(df.isnull().sum())
```

```python
#Drop unwanted columns
df = df.drop(columns=['house_rules','license'], errors Hello = 'ignore')
```

```python
df.head()
```

```python
#Remove dollar sign and conver to float for price and service fee column values
df['price'] = df['price'].replace('[\$,]','', regex = True).astype(float)
df['service fee'] = df['service fee'].replace('[\$,]','', regex = True).astype(float)
```

```python
df.head()
```

##### Remove duplicates

```python
df.drop_duplicates(inplace=True)
```
