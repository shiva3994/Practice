##### Handling missing values

```python
df['last review'] = pd.to_datetime(df['last review'],errors='coerce')
#Select columns for change
#Tell panda we wish to change it 'last review' to date time instead of str
#Errors = 'coerce' tells that if there is missing value it should is be considered as NAT format (Not a thing)
```

```python
df.head()
```

```python
df.info()
#To check the data type of 'last review'
```

```python
#Fill 'reviews per month' with 0 and replace 'last review' with minimum value of last review
df.fillna({'reviews per month' : 0, 'last review' : df['last review'].min()}, inplace=True)
```

```python
#Drop na from selected columns
df.dropna(subset = ['NAME','host name'], inplace = True)
```

```python
print(df.isnull().sum())
```

```python
#Drop unwanted columns
df = df.drop(columns=['house_rules','license'], errors = 'ignore')
```

```python
df.head()
```

```python
# #Remove dollar sign and conver to float for price and service fee column values

# Remove '$' and ',' then convert to float
# Line 1: Remove Dollar Signs
df['price'] = df['price'].astype(str).str.replace('$', '', regex=False)
# Line 2: Remove Commas
df['price'] = df['price'].str.replace(',', '', regex=False)
# Line 3: Convert Text to Numbers
df['price'] = pd.to_numeric(df['price'], errors='coerce')

df['service fee'] = df['service fee'].replace('[\$,]','', regex = True).astype(float)

# Drop any rows where price could not be converted (NaN)
df = df.dropna(subset=['price'])
```

##### Remove duplicates

```python
df.drop_duplicates(inplace=True)
```

##### Visualtsation

```python
#Distribution of listing price
plt.figure(figsize=(10, 6))

# Plotting with the cleaned numeric column
sns.histplot(data=df, x='price', bins=50, kde=True, color='orange')

plt.title('Distribution of Listing Price')
plt.xlabel('Price ($)')
plt.ylabel('Frequency')
plt.show()
```

##### How different room types are distributed

```python
df['room type']
```

```python
plt.figure(figsize=(8,5))

sns.countplot( x = 'room type', data = df, color = 'Orange')

plt.title('Room type distribution')
plt.xlabel('Room type')
plt.ylabel('Count')
plt.xticks(rotation = 50, ha = 'right')

plt.show()
```

##### Listing distribution across diff neighbourhoods.

```python
plt.figure(figsize=(12,5))

sns.countplot( y = 'neighbourhood group', data = df, color = 'Orange', order = df['neighbourhood group'].value_counts().index)

plt.title('Neighborhood distribution')
plt.xlabel('Neighbourhood')
plt.ylabel('Count')
plt.xticks(rotation=25, ha='right')
plt.show()
```

##### Relationship between price and room type

```python
plt.figure(figsize = (10,6))

sns.boxplot( x = 'room type', y = 'price', data = df , color = 'orange')

plt.title('Room type vs Price')
plt.xlabel('Room type')
plt.ylabel('Price')
plt.xticks(rotation = 30, ha = 'right')

plt.show()
```

##### How has the number of review change over with time

```python
df['last review'] = pd.to_datetime(df['last review'])

review_over_time = df.groupby(df['last review'].dt.to_period('M')).size()

plt.figure(figsize = (10,6))
review_over_time.plot(kind = 'line', color = 'orange')

plt.title('Number of review over time')
plt.xlabel('Date')
plt.ylabel('Number of reviews')

plt.show()
```
