##### Ques : What are the top 10 highest-rated iPhones on Flipkart in India?

```python
df.head()
```

```python
highest_rated = df.sort_values( by = ['Star Rating'], ascending = False)

highest_rated = highest_rated.head(10)

print(highest_rated['Product Name'])
```

##### Ans : APPLE iPhone 11 Pro Max (Gold, 64 GB)

#### **

##### Ques : How many ratings do the highest-rated iPhones on Flipkart have?

```python
df.head()
```

```python
iphones = highest_rated['Product Name'].value_counts()
label = iphones.index
counts = highest_rated["Number Of Ratings"]
figure = px.bar(highest_rated, x = label, y = counts)
figure.show()
```

##### **

##### Ques : Which iPhone has the highest number of reviews on Flipkart?

```python
df.head()
```

```python
iphones = highest_rated['Product Name'].value_counts()
label = iphones.index
counts = highest_rated["Number Of Reviews"]
figure = px.bar(highest_rated, x = label, y = counts, title = 'Number of ratings of highest rated iPhones')
figure.show()
```

##### Ques : What is the relationship between the sale price of iPhones and the number of ratings on Flipkart?

```python
df.head()
```

```python
figure = px.scatter(data_frame = df, x = 'Number Of Ratings', 
                    y = 'Sale Price', 
                    size = 'Discount Percentage', 
                    trendline = 'ols', 
                    title = 'Relationship between sales price and number of rating')

figure.show()
```

##### Ques : Can you figure out the least expensive and most expensive iPhones in the Indian market, along with all their specifications?

```python
df.head()
```

```python
most_expensive = df.loc[df['Sale Price'].idxmax()]
least_expensive = df.loc[df['Sale Price'].idxmin()]

# Display the result
print('Most Expensive procuct:')
print(most_expensive)

print('\nLeast Expensive procuct:')
print(least_expensive)
```
